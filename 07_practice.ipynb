# ebola 데이터 집합 살펴보기

```python
ebola = pd.read_csv('/content/drive/MyDrive/KUGGLE/데이터 분석을 위한 판다스 입문/data/country_timeseries.csv')
print(ebola.columns)
```

```python
print(ebola.iloc[:5, [0, 1, 2, 3, 10, 11]])
```

```python
ebola_long = pd.melt(ebola, id_vars=['Date', 'Day'])
print(ebola_long.head())
```

# 열 이름 나누고 데이터 프레임에 추가하기

```python
variable_split = ebola_long.variable.str.split('_')

print(variable_split[:5])
```

```python
print(type(variable_split))
print(type(variable_split[0]))
```

```python
status_values = variable_split.str.get(0)
country_values = variable_split.str.get(1)

print(status_values[:5])
print(status_values[-5:])
```

```python
print(country_values[:5])
print(country_values[-5:])
```

```python
ebola_long['status'] = status_values
ebola_long['country'] = country_values
print(ebola_long.head())
```

# concat 메서드를 응용하여 데이터프레임에 열 추가하기

```python
variable_split = ebola_long.variable.str.split('_', expand=True)
variable_split.columns = ['status', 'country']
ebola_parsed = pd.concat([ebola_long, variable_split], axis=1)

print(ebola_parsed.head())
```

# 기상 데이터의 여러 열을 하나로 정리하기

```python
weather = pd.read_csv('/content/drive/MyDrive/KUGGLE/데이터 분석을 위한 판다스 입문/data/weather.csv')
print(weather.iloc[:5, :])
```

```python
weather_melt = pd.melt(weather, id_vars=['id', 'year', 'month', 'element'], var_name='day', value_name='temp')
print(weather_melt.head())
```

```python
weather_tidy = weather_melt.pivot_table(
    index=['id', 'year', 'month', 'day'],
    columns='element',
    values='temp'
)

print(weather_tidy)
```

```python
weather_tidy_flat = weather_tidy.reset_index()
print(weather_tidy_flat.head())
```

# 뉴욕 택시 데이터 준비

```python
import os
import urllib.request

# 네트워크 상태에 따라 5 ~ 15분이 소요됩니다.
with open('../data/raw_data_urls.txt', 'r') as data_urls:
    for line, url in enumerate(data_urls):
        if line == 5:
            break
        fn = url.split('/')[-1].strip()
        fp = os.path.join('', '../data', fn)
        print(url)
        print(fp)
        urllib.request.urlretrieve(url, fp)
```

```python
import glob
nyc_taxi_data = glob.glob('/content/drive/MyDrive/KUGGLE/데이터 분석을 위한 판다스 입문/data/fhv_*')
print(nyc_taxi_data)
```

```python
taxi1 = pd.read_csv(nyc_taxi_data[0])
taxi2 = pd.read_csv(nyc_taxi_data[1])
taxi3 = pd.read_csv(nyc_taxi_data[2])
taxi4 = pd.read_csv(nyc_taxi_data[3])
taxi5 = pd.read_csv(nyc_taxi_data[4])
```

```python
print(taxi1.shape)
print(taxi2.shape)
print(taxi3.shape)
print(taxi4.shape)
print(taxi5.shape)
```

```python
taxi = pd.concat([taxi1, taxi2, taxi3, taxi4, taxi5])

print(taxi.shape)
```

# 반복문으로 데이터 준비하기

```python
list_taxi_df = []

for csv_filename in nyc_taxi_data:
    # print(csv_filename)
    df = pd.read_csv(csv_filename)
    list_taxi_df.append(df)

print(len(list_taxi_df))
```

```python
print(type(list_taxi_df[0]))

print(list_taxi_df[0].head())
```

```python
taxi_loop_concat = pd.concat(list_taxi_df)
print(taxi_loop_concat.shape)

print(taxi.equals(taxi_loop_concat))
```
