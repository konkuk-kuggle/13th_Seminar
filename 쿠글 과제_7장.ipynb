<concat 메서드를 응용하여 데이터프레임에 열 추가하기>

```python
variable_split = ebola_long.variable.str.split('_', expand=True)
variable_split.columns = ['status', 'country']
ebola_parsed = pd.concat([ebola_long, variable_split], axis=1)
#split으로 분리, concat으로 열 추가
print(ebola_parsed.head())
```

기상 데이터의 여러 열을 하나로 정리하기 ─ melt, pivot_table 메서드>

```python
weather = pd.read_csv('../data/weather.csv')
print(weather.iloc[:5, :11]) #날짜 열을 행데이터로 피벗
```

```python
weather_melt = pd.melt(weather, id_vars=['id', 'year', 'month', 'element'], var_name='day', value_name='temp')
#날짜열의 데이터 temp열에 정리
print(weather_melt.head())
```

```python
weather_tidy = weather_melt.pivot_table(
    index=['id', 'year', 'month', 'day'], #위치 유지할 열
    columns='element', #피벗할 열
    values='temp' #새로운 열의 데이터가 될 열 이름
)

print(weather_tidy)
```

```python
weather_tidy_flat = weather_tidy.reset_index()
#데이터프레임의 인덱스를 reset_index로 새로 지정
print(weather_tidy_flat.head())
```

<빌보드 차트의 중복 데이터 처리하기>

```python
billboard = pd.read_csv('../data/billboard.csv')
billboard_long = pd.melt(billboard, id_vars=['year', 'artist', 'track', 'time', 'date.entered'], var_name='week', value_name='rating')

print(billboard_long.shape)

print(billboard_long.head())
```

```python
print(billboard_long[billboard_long.track == 'Loser'].head())
#노래 제목이 Loser인 데이터만 모으기
```

```python
billboard_songs = billboard_long[['year', 'artist', 'track', 'time']]
#중복데이터 가진 열 모으기
print(billboard_songs.shape)
```

```python
billboard_songs = billboard_songs.drop_duplicates()
#drop_duplicates로 중복 제거
print(billboard_songs.shape)
```

```python
billboard_songs['id'] = range(len(billboard_songs)) #아이디 추가
print(billboard_songs.head(n=10))
```

```python
billboard_ratings = billboard_long.merge( billboard_songs, on=['year', 'artist', 'track', 'time'])
print(billboard_ratings.shape)
#merge로 주간 순위 데이터 합치기
```

```python
print(billboard_ratings.head())
```

```python
import os
import urllib.request

# 네트워크 상태에 따라 5 ~ 15분이 소요됩니다.
with open('../data/raw_data_urls.txt', 'r') as data_urls:
    for line, url in enumerate(data_urls):
        if line == 5:
            break
        fn = url.split('/')[-1].strip()
        fp = os.path.join('', '../data', fn)
        print(url)
        print(fp)
        urllib.request.urlretrieve(url, fp)
```

```python
import glob
nyc_taxi_data = glob.glob('../data/fhv_*')
print(nyc_taxi_data)
```

```python
#각각의 파일을 데이터프레임으로 저장하기
taxi1 = pd.read_csv(nyc_taxi_data[0])
taxi2 = pd.read_csv(nyc_taxi_data[1])
taxi3 = pd.read_csv(nyc_taxi_data[2])
taxi4 = pd.read_csv(nyc_taxi_data[3])
taxi5 = pd.read_csv(nyc_taxi_data[4])
```

<반복문으로 데이터 준비하기>

```python
list_taxi_df = []
#반복문으로 파일 목록을 읽어서 리스트에 이어붙이기
for csv_filename in nyc_taxi_data:
    # print(csv_filename)
    df = pd.read_csv(csv_filename)
    list_taxi_df.append(df)

print(len(list_taxi_df))
```

```python
print(list_taxi_df[0].head())
```

```python
taxi_loop_concat = pd.concat(list_taxi_df)
print(taxi_loop_concat.shape)
```
