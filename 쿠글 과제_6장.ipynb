<누락값이 생기는 이유 알아보기>

```python
visited = pd.read_csv('../data/survey_visited.csv')
survey = pd.read_csv('../data/survey_survey.csv')

print(visited)
```

3. 데이터를 입력할 때 누락값이 생기는 경우

```python
num_legs = pd.Series({'goat': 4, 'amoeba': nan})
print(num_legs)
print(type(num_legs))
```

```python
scientists = pd.DataFrame({
    'Name': ['Rosaline Franklin', 'William Gosset'],
    'Occupation': ['Chemist', 'Statistician'],
    'Born': ['1920-07-25', '1876-06-13'],
    'Died': ['1958-04-16', '1937-10-16'],
    'missing': [NaN, nan]}) #누락값

print(scientists)
print(type(scientists))
```

4. 범위를 지정하여 데이터를 추출할 때 누락값이 생기는 경우

```python
gapminder = pd.read_csv('../data/gapminder.tsv', sep='\t')
```

```python
life_exp = gapminder.groupby(['year'])['lifeExp'].mean() #평균
print(life_exp)
```

```python
print(life_exp.loc[range(2000, 2010), ])
#2000~2009년의 데이터 추출
```

```python
y2000 = life_exp[life_exp.index > 2000] #불린추출
print(y2000)
```

<누락값의 개수 구하기>

```python
ebola = pd.read_csv('../data/country_timeseries.csv')
```

```python
print(ebola.count()) #누락값이 아닌 값의 개수 구하기
```

```python
num_rows = ebola.shape[0]
num_missing = num_rows - ebola.count() #누락값의 개수
print(num_missing)
```

```python
import numpy as np
print(np.count_nonzero(ebola.isnull()))
#count_nonzero와 isnull 메서드 조합해서 누락값 개수 구하기
```

```python
print(np.count_nonzero(ebola['Cases_Guinea'].isnull()))
```

```python
print(ebola.Cases_Guinea.value_counts(dropna=False).head())
#지정한 열의 빈도를 구하는 메서드
```

<누락값 처리하기 ― 변경, 삭제>

1. 누락값 변경하기

```python
print(ebola.fillna(0).iloc[0:10, 0:5])
#누락값을 0으로 변경
```

```python
print(ebola.fillna(method='ffill').iloc[0:10, 0:5])
#누락값이 나타나기 전의 값으로 누락값을 변경
```

```python
print(ebola.fillna(method='bfill').iloc[0:10, 0:5])
#위와 반대방향으로 누락값을 처리
```

```python
print(ebola.interpolate().iloc[0:10, 0:5])
#중간값을 구해 누락값 처리
```

5. 누락값 삭제하기

```python
print(ebola.shape) #데이터 구조 확인
```

```python
ebola_dropna = ebola.dropna() #누락값 포함 행들 모두 삭제
print(ebola_dropna.shape)
```

```python
print(ebola_dropna)
```

<누락값이 포함된 데이터 계산하기>

```python
#새로운 열 만들기
ebola['Cases_multiple'] = ebola['Cases_Guinea'] + ebola['Cases_Liberia'] + ebola['Cases_SierraLeone']
```

```python
ebola_subset = ebola.loc[:, ['Cases_Guinea', 'Cases_Liberia', 'Cases_SierraLeone', 'Cases_multiple']]
#새로운 데이터 프레임
print(ebola_subset.head(n=10))
```

```python
print(ebola.Cases_Guinea.sum(skipna = True)) #누락값 무시
#sum : 누락값 포함해 계산
```

```python
print(ebola.Cases_Guinea.sum(skipna = False))
#누락값 무시 못함
```
