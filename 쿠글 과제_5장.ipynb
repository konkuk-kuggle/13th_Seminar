<다양한 방법으로 데이터 연결하기>

1. ignore_index 인자 사용하기

```python
row_concat_i = pd.concat([df1, df2, df3], ignore_index=True)
#ignore_index 인덱스를 0부터 다시지정
print(row_concat_i)
```

2. 열 방향으로 데이터 연결하기

```python
col_concat = pd.concat([df1, df2, df3], axis=1) #열방향으로 데이터 연결
print(col_concat)
```

```python
print(col_concat['A']) #해당 열 이름의 데이터 모두 추출
```

```python
col_concat['new_col_list'] = ['n1', 'n2', 'n3', 'n4'] #새로운 열 추가
print(col_concat)
```

```python
print(pd.concat([df1, df2, df3], axis=1, ignore_index=True))
#중복 열이름->열 이름 재지정
```

6. 공통 열과 공통 인덱스만 연결하기

```python
df1.columns = ['A', 'B', 'C', 'D']
df2.columns = ['E', 'F', 'G', 'H']
df3.columns = ['A', 'C', 'F', 'H']
#열 이름 재지정
print(df1)
print(type(df1))
```

```python
print(df2)
print(type(df2))
```

```python
print(df3)
print(type(df3))
#열 이름 중복
```

```python
row_concat = pd.concat([df1, df2, df3]) #데이터프레임 연결
print(row_concat)
```

```python
print(pd.concat([df1, df2, df3], join='inner'))
#공통 열만 골라 연결하면 누락값 발생 안함
```

```python
print(pd.concat([df1,df3], ignore_index=False, join='inner'))
#교집합인 것을 추출
```

```python
df1.index = [0, 1, 2, 3]
df2.index = [4, 5, 6, 7]
df3.index = [0, 2, 5, 7]
#행 방향 연결
print(df1)
```

```python
print(df2)
```

```python
print(df3)
```

```python
col_concat = pd.concat([df1, df2, df3], axis=1)
print(col_concat)
#행 방향 연결
```

```python
print(pd.concat([df1, df3], axis=1, join='inner'))
#공통행만
```

<merge 메서드 사용하기>

```python
person = pd.read_csv('../data/survey_person.csv')
site = pd.read_csv('../data/survey_site.csv')
survey = pd.read_csv('../data/survey_survey.csv')
visited = pd.read_csv('../data/survey_visited.csv')

print(person)
```

```python
print(site)
```

```python
print(visited)
```

```python
print(survey)
```

```python
visited_subset = visited.loc[[0, 2, 6], ]
print(visited_subset)
#일부 데이터만 떼어 사용
```

```python
o2o_merge = site.merge(visited_subset, left_on='name', right_on='site')
print(o2o_merge)
```

```python
m2o_merge = site.merge(visited, left_on='name', right_on='site')
print(m2o_merge)
```

```python
ps = person.merge(survey, left_on='ident', right_on='person')
vs = visited.merge(survey, left_on='ident', right_on='taken')

print(ps)
```

```python
print(vs)
```

```python
ps_vs = ps.merge(vs, left_on=['ident', 'taken', 'quant', 'reading'], right_on=['person', 'ident', 'quant', 'reading'])
```

```python
print(ps_vs.loc[0, ])
#중복 열 이름에 접미사 추가됨
```
