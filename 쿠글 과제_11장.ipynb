<평균값을 구하는 사용자 함수와 groupby 메서드>

```python
def my_mean(values):
    n = len(values)

    sum = 0
    for value in values:
        sum += value

    return sum / n

#평균값 구하는 함수
```

```python
agg_my_mean = df.groupby('year').lifeExp.agg(my_mean)
#agg 메서드 사용 값 = mean 메서드 사용 값
print(agg_my_mean)
```

<두 개의 인잣값을 받아 처리하는 사용자 함수와 groupby 메서드>

```python
def my_mean_diff(values, diff_value):
    n = len(values)
    sum = 0
    for value in values:
        sum += value
    mean = sum / n
    return mean - diff_value
#평균값 구하고 두번째 인자로 받은 값과의 차이 계산
```

```python
global_mean = df.lifeExp.mean()
print(global_mean)
#전체수명
```

```python
agg_mean_diff = df.groupby('year').lifeExp.agg(my_mean_diff, diff_value=global_mean)
print(agg_mean_diff)
#연도별 평균 수명-전체 평균 수명
```

<표준점수 계산하기>

```python
def my_zscore(x):
    return (x - x.mean()) / x.std()
    #표준점수 계산
```

```python
transform_z = df.groupby('year').lifeExp.transform(my_zscore)
#연도별 표준점수 계산
print(transform_z.head())
```

<누락값을 평균값으로 처리하기>

```python
import seaborn as sns
import numpy as np

np.random.seed(42)
tips_10 = sns.load_dataset('tips').sample(10)
tips_10.loc[np.random.permutation(tips_10.index)[:4], 'total_bill'] = np.NaN
#total_bill 열의 값 4개를 임의선택해 누락값으로 바꾸기
print(tips_10)
```

```python
count_sex = tips_10.groupby('sex').count()
print(count_sex)
#성별로 그룹화 후 각 열의 데이터 수 구하기
```

```python
def fill_na_mean(x):
    avg = x.mean()
    return x.fillna(avg) #평균값 구하기
```

```python
total_bill_group_mean = tips_10.groupby('sex').total_bill.transform(fill_na_mean)
tips_10['fill_total_bill'] = total_bill_group_mean
print(tips_10)
#누락값을 고려해 계산한 평균값이 fill_total_bill에 채워짐
```
