# 시리즈 만들기

```python
import pandas as pd
s=pd.Series(['banana',42])
print(s)

#문자열을 인덱스로 지정
s=pd.Series(['Wes McKinney','Creator of Pandas'])
print(s)

s=pd.Series(['Wes McKinney','Creator of Pandas'],index=['Person','Who'])
print(s)
```

# 데이터프레임 만들기

```python
scientists=pd.DataFrame({
    'Name':['Rosaline Franklin','William Gosset'],
    'Occupation':['Chemist','Statistician'],
    'Born':['1920-07-25','1876-06-13'],
    'Died':['1958-04-16','1937-10-16'],
    'Age':[37,61]}
                       )
print(scientists)

#인덱스 지정하기, 열 순서 지정하기
scientists=pd.DataFrame(
    data={'Occupation':['Chemist','Statistician'],
          'Born':['1920-07-25','1876-06-13'],
          'Died':['1958-04-16','1937-10-16'],
          'Age':[37,61]},
    index=['Rosaline Franklin','William Gosset'],
    columns=['Occupation','Born','Age','Died']
)
print(scientists)

#딕셔너리는 키(Key)와 값(Value)으로 이루어진 한 쌍의 데이터들의 순서를 보장하지 않음-> OrderedDict로 순서보장
from collections import OrderedDict

scientists=pd.DataFrame(OrderedDict([
    ('Name',['Rosaline Franklin','William Gosset']),
    ('Occupation',['Chemist','Statistician']),
    ('Born',['1920-07-25','1876-06-13']),
    ('Died',['1958-04-16','1937-10-16']),
    ('Age',[37,61])
])
)
print(scientists)

    
```

# 데이터프레임에서 시리즈 선택하기

```python
scientists=pd.DataFrame(
    data={'Occupation':['Chemist','Statistician'],
          'Born':['1920-07-25','1876-06-13'],
          'Died':['1958-04-16','1937-10-16'],
          'Age':[37,61]},
    index=['Rosaline Franklin','William Gosset'],
    columns=['Occupation','Born','Died','Age'])

first_row = scientists.loc['William Gosset']
print(type(first_row))

print(first_row)
```

# index, values, keys 사용하기

```python
#index 속성 사용하기
print(first_row.index)

#values 속성 사용하기
print(first_row.values)

#keys 메서드 사용하기
print(first_row.keys())

#인덱스 첫 번째 값 추출
print(first_row.index[0])
print(first_row.keys()[0])


```

# 시리즈의 mean, min, max, std 메서드 사용하기

```python
ages=scientists['Age']
print(ages)

print(ages.mean())
print(ages.min())
print(ages.max())
```

```python
#불린 추출: 특정 조건을 만족하는 값만 추출
scientists=pd.read_csv('../data/scientists.csv')

ages=scientists['Age']
print(ages.max())
print(ages.mean())

print(ages[ages>ages.mean()])

print(ages>ages.mean())
print(type(ages>ages.mean()))

manual_bool_values=[True,True,False,False,True,True,False,True]
print(ages[manual_bool_values])
```

# 시리즈와 브로드캐스팅

```python
#브로드캐스팅: 시리즈나 데이터프레임에 있는 모든 데이터에 대해 한 번에 연산하는 것

print(ages+ages)
print(ages*ages)
print(ages+100)
print(ages*2)

#길이가 서로 다른 벡터 연산
print(pd.Series([1,100]))
print(ages+pd.Series([1,100]))

rev_ages=ages.sort_index(ascending=False)    #인덱스 역순으로 데이터 정렬
print(ages)
print(rev_ages)

print(ages*2)
print(ages+rev_ages) #ages*2와 결과값이 동일 (벡터와 벡터 연산은 일치하는 인덱스 값끼리 수행)
```

# 열의 자료형 바꾸기와 새로운 열 추가하기

```python
#열의 자료형 바꾸기
print(scientists['Born'].dtype)
print(scientists['Died'].dtype)

born_datetime=pd.to_datetime(scientists['Born'],format='%Y-%m-%d')
print(born_datetime)

died_datetime=pd.to_datetime(scientists['Died'],format='%Y-%m-%d')
print(died_datetime)

#새로운 열 추가하기
scientists['born_dt'],scientists['died_dt']=(born_datetime,died_datetime)
print(scientists.head())

print(scientists.shape)

#시간 계산하기
scientists['age_days_dt']=(scientists['died_dt']-scientists['born_dt'])
print(scientists)
```
