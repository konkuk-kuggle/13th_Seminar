<제곱 함수와 n 제곱 함수 만들기>

```python
def my_sq(x):
    return x ** 2 #제곱함수
```

```python
def my_exp(x, n):
    return x ** n #n제곱함수
```

```python
print(my_exp(2, 4))
```

```python
def avg_3(x, y, z):
    return (x + y + z) / 3 #평균 계산 함수
```

```python
print(df.apply(avg_3)) #인자를 한개만 주면 오류
```

```python
def avg_3_apply(col):
    x = col[0]
    y = col[1]
    z = col[2]
    return (x + y + z) / 3

#개선한 함수를 apply 메서드에 적용
print(df.apply(avg_3_apply))
```

```python
def avg_3_apply(col):
    sum = 0
    for item in col: #for문 사용
         sum += item
    return sum / df.shape[0]
```

```python
def avg_2_apply(row): #행방향으로 처리
    sum = 0
    for item in row:
        sum += item
    return sum / df.shape[1] #1을 사용
```

```python
print(df.apply(avg_2_apply, axis = 1))
```

<데이터프레임의 누락값을 처리한 다음 apply 메서드 사용하기>

1. 데이터프레임의 누락값 처리하기 ― 열 방향

```python
import seaborn as sns

titanic = sns.load_dataset("titanic")
```

```python
print(titanic.info())
```

```python
import numpy as np

def count_missing(vec):
    null_vec = pd.isnull(vec)
    null_count = np.sum(null_vec)
    return null_count
```

```python
cmis_col = titanic.apply(count_missing) #누락값 세기
print(cmis_col)
```

```python
def prop_missing(vec):
    num = count_missing(vec)
    dem = vec.size
    return num / dem #누락값의 비율 계산
```

```python
pmis_col = titanic.apply(prop_missing)
print(pmis_col)
```

```python
def prop_complete(vec):
    return 1 - prop_missing(vec) #데이터의 비율=1-누락값의 비율
```

8. 데이터프레임의 누락값을 처리하기 ― 행 방뱡

```python
cmis_row = titanic.apply(count_missing, axis=1) #행 방향
pmis_row = titanic.apply(prop_missing, axis=1) #행 방향
pcom_row = titanic.apply(prop_complete, axis=1) #행 방향

print(cmis_row.head())
```

```python
print(pmis_row.head())
```

```python
print(pcom_row.head())
```

```python
titanic['num_missing'] = titanic.apply(count_missing, axis=1)
#num_missing 열 추가
print(titanic.head())
```

```python
print(titanic.loc[titanic.num_missing > 1, :].sample(10))
#누락값이 2개 이상인 데이터 추출
```
