# 자료형을 자유자재로 변환하기-astype 메서드

```python
import pandas as pd
import seaborn as sns

tips = sns.load_dataset("tips")
```

# 여러 가지 자료형을 문자열로 변환하기

```python
tips['sex_str'] = tips['sex'].astype(str)
print(tips.dtypes)
```

# 자료형을 변환한 데이터 다시 원래대로 만들기

```python
tips['total_bill'] = tips['total_bill'].astype(str) 
print(tips.dtypes)

tips['total_bill'] = tips['total_bill'].astype(float) #다시 실수로 변환
print(tips.dtypes)
```

# 잘못 입력한 문자열 처리하기

```python
#to_numeric 메서드

tips_sub_miss = tips.head(10)
tips_sub_miss.loc[[1, 3, 5, 7], 'total_bill'] = 'missing'
print(tips_sub_miss)

print(tips_sub_miss.dtypes)  #total_bill 열 자료형: 문자열

#오류 tips_sub_miss['total_bill'].astype(float)
#오류 pd.to_numeric(tips_sub_miss['total_bill'])
'''to_numeric의 erros인자
raise: 숫자로 변환할 수 없는 값이 있으면 오류 발생
coerce: 숫자로 변환할 수 없는 값을 누락값으로 지정
ignore: 아무 작업도 하지 않음'''

tips_sub_miss['total_bill'] = pd.to_numeric( tips_sub_miss['total_bill'], errors='ignore')
print(tips_sub_miss.dtypes)

tips_sub_miss['total_bill'] = pd.to_numeric( tips_sub_miss['total_bill'], errors='coerce')
print(tips_sub_miss.dtypes)
#print(tips_sub_miss['total_bill'])

#to_numeric의 downcast인자 : 정수, 실수와 같은 자료형을 더 작은 형태로  (저장하는 실수의 예상 범위가 크지 않을 때 메모리 공간 절약)
tips_sub_miss['total_bill'] = pd.to_numeric( tips_sub_miss['total_bill'], errors='coerce', downcast='float')
print(tips_sub_miss.dtypes)
```

# 문자열을 카테고리로 변환하기

```python
#카테고리 자료형: 유한한 범위의 값만을 가짐
tips['sex'] = tips['sex'].astype('str') 
print(tips.info())

tips['sex'] = tips['sex'].astype('category') 
print(tips.info())
```
