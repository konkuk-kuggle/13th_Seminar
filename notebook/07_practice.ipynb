# 1개의 열만 고정하고 나머지 열을 행으로 바꾸기

```python
import pandas as pd
pew = pd.read_csv('../data/pew.csv')
print(pew.head())
print(pew.iloc[:, 0:6])
```

```python
#melt 메서드 : 지정한 열의 데이터를 모두 행으로 정리
pew_long = pd.melt(pew, id_vars='religion') #id_vars : 위치를 그대로 유지할 열의 이름
print(pew_long.head())

#var_name : value_vars로 위치를 변경한 열의 이름, value_name : var_name으로 위치를 변경할 열의 데이터를 저장한 열의 이름
pew_long = pd.melt(pew, id_vars='religion', var_name='income', value_name='count')
print(pew_long.head())
```

# ebola 데이터 집합 살펴보기

```python
ebola = pd.read_csv('../data/country_timeseries.csv')
print(ebola.columns)
print(ebola.iloc[:5, [0, 1, 2, 3, 10, 11]])

ebola_long = pd.melt(ebola, id_vars=['Date', 'Day']) #Date, Day 고정하여 피벗
print(ebola_long.head())
```

# 열 이름 나누고 데이터 프레임에 추가하기

```python
variable_split = ebola_long.variable.str.split('_') #'_' 문자를 기준으로 분리
print(variable_split[:5]) #Cases_Guinea -> Cases, Guinea
print(type(variable_split))
print(type(variable_split))

status_values = variable_split.str.get(0) #발병, 죽음 (Cases)
country_values = variable_split.str.get(1) #나라 이름 (Guinea)
print(status_values[:5])
print(status_values[-5:])
print(country_values[:5])
print(country_values[-5:])

#분리한 문자열로 데이터프레임에 추가
ebola_long['status'] = status_values 
ebola_long['country'] = country_values
print(ebola_long.head())
```

# concat 메서드를 응용하여 데이터프레임에 열 추가하기

```python
variable_split = ebola_long.variable.str.split('_', expand=True) 
variable_split.columns = ['status', 'country'] 
ebola_parsed = pd.concat([ebola_long, variable_split], axis=1) #concat() : 분리한 데이터를 바로 데이터프레임에 추가

print(ebola_parsed.head())
```

# 기상 데이터의 여러 열을 하나로 정리하기

```python
weather = pd.read_csv('../data/weather.csv') 
print(weather.iloc[:5, :11])
```

```python
weather_melt = pd.melt(weather, id_vars=['id', 'year', 'month', 'element'], var_name='day', value_name='temp') 
print(weather_melt.head()) #일별 온도(d1~...) 피벗

weather_tidy = weather_melt.pivot_table( #pivot_table : 행과 열의 위치를 바꿔 정리
    index=['id', 'year', 'month', 'day'], #위치를 유지할 열 이름
    columns='element', #피벗할 열 이름
    values='temp' #새로운 열의 데이터가 될 열 이름
)
print(weather_tidy)
```

```python
weather_tidy_flat = weather_tidy.reset_index() #데이터프레임 인덱스를 reset_index로 새로 지정(0~)
print(weather_tidy_flat.head())
```

# 빌보드 차트의 중복 데이터 처리하기

```python
billboard = pd.read_csv('../data/billboard.csv')
billboard_long = pd.melt(billboard, id_vars=['year', 'artist', 'track', 'time', 'date.entered'], var_name='week', value_name='rating')
print(billboard_long.shape)
print(billboard_long.head())

#중복데이터를 가지고 있는 열
print(billboard_long[billboard_long.track == 'Loser'].head()) #노래제목이 Loser인 곡이 많음
billboard_songs = billboard_long[['year', 'artist', 'track', 'time']]
print(billboard_songs.shape)

billboard_songs = billboard_songs.drop_duplicates() #drop_duplicates : 중복 데이터 제거
print(billboard_songs.shape)

billboard_songs['id'] = range(len(billboard_songs)) #중복을 제거한 데이터 프레임에 id 추가
print(billboard_songs.head(n=10))

#billboard_long을 기준으로 billboard_songs 데이터를 합침
billboard_ratings = billboard_long.merge( billboard_songs, on=['year', 'artist', 'track', 'time']) 
print(billboard_ratings.shape)
print(billboard_ratings.head())
```

# 뉴욕 택시 데이터 준비

```python
#새로운 주소(데이터 주소가 바뀜)
base_url = "https://d37ci6vzurychx.cloudfront.net/trip-data/fhv_tripdata_2015-"
months = ["01", "02", "03", "04", "05"]

#파일 쓰기 ('w' : 기존 내용을 지우고 새로 작성)
with open('../data/raw_data_urls.txt', 'w') as f:
    for month in months:
        new_url = f"{base_url}{month}.parquet\n"
        f.write(new_url)
```

```python
import os 
import urllib.request

with open('../data/raw_data_urls.txt', 'r') as data_urls:
    for line, url in enumerate(data_urls):
        if line == 5:
            break 
        fn = url.split('/')[-1].strip()
        fp = os.path.join('', '../data', fn)
        print(url)
        print(fp)
        urllib.request.urlretrieve(url, fp)
```

```python
#데이터 불러오기
import glob 
nyc_taxi_data = glob.glob('../data/fhv_*') #glob() : 특정한 패턴의 이름을 가진 파일을 한 번에 읽어옴
print(nyc_taxi_data)
```

```python
taxi1 = pd.read_parquet(nyc_taxi_data[0]) #2025-01
taxi2 = pd.read_parquet(nyc_taxi_data[1]) #2025-02
taxi3 = pd.read_parquet(nyc_taxi_data[2]) #2025-03
taxi4 = pd.read_parquet(nyc_taxi_data[3]) #2025-04
taxi5 = pd.read_parquet(nyc_taxi_data[4]) #2025-05

print(taxi1.head(n=2)) 
print(taxi2.head(n=2)) 
print(taxi3.head(n=2)) 
print(taxi4.head(n=2)) 
print(taxi5.head(n=2))

print(taxi1.shape) 
print(taxi2.shape) 
print(taxi3.shape) 
print(taxi4.shape) 
print(taxi5.shape)

taxi = pd.concat([taxi1, taxi2, taxi3, taxi4, taxi5]) #모든 데이터프레임 연결

print(taxi.shape)
```

# 반복문으로 데이터 준비하기

```python
list_taxi_df = [] 
for csv_filename in nyc_taxi_data: #생성한 파일 목록(nyc_taxi_data)을 반복문으로 읽어 들임
    # print(csv_filename)
    df = pd.read_parquet(csv_filename)
    list_taxi_df.append(df) #읽어들인 파일을 리스트에 이어붙임

print(len(list_taxi_df))
print(type(list_taxi_df[0]))
print(list_taxi_df[0].head())

taxi_loop_concat = pd.concat(list_taxi_df) 
print(taxi_loop_concat.shape)
print(taxi.equals(taxi_loop_concat)) #반복문으로 저장한 데이터와 concat으로 저장한 데이터가 같은지 비교
```
