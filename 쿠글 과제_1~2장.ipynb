<그룹화한 데이터의 평균 구하기>
1. lifeExp 열을 연도별로 그룹화하여 평균 계산하기

```python
print(df.head(n=10))
```

```python
print(df.groupby('year')['lifeExp'].mean()) #연도별로 그룹화하여 평균 계산
```

```python
grouped_year_df = df.groupby('year')
print(type(grouped_year_df))
```

```python
print(grouped_year_df) #메모리위치
```

```python
grouped_year_df_lifeExp = grouped_year_df['lifeExp']
print(type(grouped_year_df_lifeExp))
```

```python
mean_lifeExp_by_year = grouped_year_df_lifeExp.mean()
print(mean_lifeExp_by_year)
```

6. lifeExp, gdpPercap 열의 평균값을 연도, 지역별로 그룹화하여 한 번에 계산하기

```python
multi_group_var = df.groupby(['year', 'continent'])[['lifeExp', 'gdpPercap']].mean()
print(multi_group_var)
```

```python
print(type(multi_group_var))
```

7. 그룹화한 데이터의 개수 세기

```python
print(df.groupby('continent')['country'].nunique()) #데이터의 빈도수
```
