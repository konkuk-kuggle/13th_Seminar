# 범위를 지정하여 데이터를 추출할 때 누락값이 생기는 경우

```python
gapminder = pd.read_csv('/content/drive/MyDrive/KUGGLE/데이터 분석을 위한 판다스 입문/data/gapminder.tsv', sep='\t')
```

```python
life_exp = gapminder.groupby(['year'])['lifeExp'].mean()
print(life_exp)
```

```python
print(life_exp.loc[range(2000, 2010), ])
```

```python
y2000 = life_exp[life_exp.index > 2000]
print(y2000)
```

# 누락값의 개수 구하기

```python
ebola = pd.read_csv('/content/drive/MyDrive/KUGGLE/데이터 분석을 위한 판다스 입문/data/country_timeseries.csv')
```

```python
print(ebola.count())
```

```python
num_rows = ebola.shape[0]
num_missing = num_rows - ebola.count()
print(num_missing)
```

```python
import numpy as np

print(np.count_nonzero(ebola.isnull()))
```

```python
print(np.count_nonzero(ebola['Cases_Guinea'].isnull()))
```

```python
print(ebola.Cases_Guinea.value_counts(dropna=False).head())
```

# 누락값 변경하기

```python
print(ebola.fillna(0).iloc[0:10, 0:5])
```

```python
print(ebola.fillna(method='ffill').iloc[0:10, 0:5])
```

```python
print(ebola.fillna(method='bfill').iloc[0:10, 0:5])
```

```python
print(ebola.interpolate().iloc[0:10, 0:5])
```
