# 누락값 확인하기

```python
from numpy import nan as NaN

print(NaN == True)
print(NaN == False)
print(NaN == 0)
print(NaN == '')

print(NaN == NaN)
#print(NaN == nan)  #오류
#print(NaN == NAN)  #오류
#print(nan == NAN)  #오류


import pandas as pd 
#누락값 확인 메서드 isnull
print(pd.isnull(NaN))
#print(pd.isnull(nan))  #오류
#print(pd.isnull(NAN))  #오류
print(pd.notnull(NaN))
print(pd.notnull(42))
print(pd.notnull('missing'))
```

# 누락값을 포함한 데이터를 불러올 때

```python
visited = pd.read_csv('../data/survey_visited.csv') 
survey = pd.read_csv('../data/survey_survey.csv')
print(visited)
print(survey)
```

# 데이터 집합을 연결할 때 누락값이 발생하는 경우

```python
vs = visited.merge(survey, left_on='ident', right_on='taken') 
print(vs)
```

# 데이터를 입력할 때 누락값이 발생하는 경우

```python
num_legs = pd.Series({'goat': 4, 'amoeba': NaN}) 
print(num_legs)
print(type(num_legs))

scientists = pd.DataFrame({ 
    'Name': ['Rosaline Franklin', 'William Gosset'], 
    'Occupation': ['Chemist', 'Statistician'], 
    'Born': ['1920-07-25', '1876-06-13'], 
    'Died': ['1958-04-16', '1937-10-16'], 
    'missing': [NaN, NaN]}) 

print(scientists)
print(type(scientists))
```

# 범위를 지정하여 데이터를 추출할 때 누락값이 발생하는 경우

```python
gapminder = pd.read_csv('../data/gapminder.tsv', sep='\t')

life_exp = gapminder.groupby(['year'])['lifeExp'].mean() 
print(life_exp)

#print(life_exp.loc[range(2000,2010),])  #오류
print(life_exp.loc[2000:2010])

#불린 추출로 range메서드 사용할 떄 발생하는 누락값 발생 문제 해결
y2000 = life_exp[life_exp.index > 2000] 
print(y2000)
```

# 누락값의 개수 구하기

```python
ebola = pd.read_csv('../data/country_timeseries.csv')

#누락값이 아닌 값의 개수
print(ebola.count())

#누락값의 개수_sol1(shape, count)
num_rows = ebola.shape[0] #전체 행의 데이터 개수
num_missing = num_rows - ebola.count() 
print(num_missing)

#누락값의 개수_sol2(count_nonzero, isnull)
import numpy as np 
print(np.count_nonzero(ebola.isnull()))

print(np.count_nonzero(ebola['Cases_Guinea'].isnull()))

#누락값의 개수_sol3(value_counts(dropna=False): 비어있는 값(NaN)도 하나의 종류로 취급해서 빈도 구하기)
print(ebola.Cases_Guinea.value_counts(dropna=False).head())
```

# 누락값을 다른 값으로 변경하기

```python
print(ebola.fillna(0).iloc[0:10, 0:5]) #누락값을 0으로
print(ebola.fillna(method='ffill').iloc[0:10, 0:5]) #누락값을 누락값이 나타나기 전 값으로 ->0,1행은 누락값 그대로
print(ebola.fillna(method='bfill').iloc[0:10, 0:5]) #누락값을 누락값이 나타난 이후의 첫 번째 값으로 ->마지막 값이 누락값인 경우 누락값 그대로
print(ebola.interpolate().iloc[0:10, 0:5]) #누락값을 누락값 양쪽 값의 중간값으로
```

# 누락값 삭제하기

```python
print(ebola.shape)

ebola_dropna = ebola.dropna() 
print(ebola_dropna.shape)

print(ebola_dropna)
```

# 누락값이 포함된 데이터 계산하기

```python
ebola['Cases_multiple'] = ebola['Cases_Guinea'] + ebola['Cases_Liberia'] + ebola['Cases_SierraLeone']

ebola_subset = ebola.loc[:, ['Cases_Guinea', 'Cases_Liberia', 'Cases_SierraLeone', 'Cases_multiple']] 
print(ebola_subset.head(n=10))

print(ebola.Cases_Guinea.sum(skipna = True))  #누락값 무시 후 계산

print(ebola.Cases_Guinea.sum(skipna = False))
```
