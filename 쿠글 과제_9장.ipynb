<문자열 포매팅하기>

<숫자 데이터 포매팅하기>

```python
print('Some digits of pi: {}'.format(3.14159265359))
#숫자도 플레이스 홀더 사용가능
```

```python
print("In 2005, Lu Chao of China recited {:,} digits of pi".format(67890))
#쉼표로 숫자 표현가능
```

```python
print("I remember {0:.4} or {0:.4%} of what Lu Chao recited".format(7/67890))
#소수
#.4 : 소수점 이하 4개까지
# % : 백분율로 환산하여 출력
```

```python
print("My ID number is {0:05d}".format(42))
#5자리수로 표현하되 빈칸 0으로
```

<정규식으로 전화번호 패턴 찾기>

```python
import re

tele_num = '1234567890'
```

```python
m = re.match(pattern='\d\d\d\d\d\d\d\d\d\d', string=tele_num)
print(type(m))
# \d : 숫자
```

```python
print(m)
```

```python
print(bool(m))
```

```python
if m:
    print('match')
else:
    print('no match')
```

```python
print(m.start()) #첫번째 인덱스 반환
```

```python
print(m.end()) #마지막 인덱스 반환
```

```python
print(m.span()) #찾은 패턴의 첫,마지막 인덱스 반환
```

```python
print(m.group()) #찾아낸 패턴 반환
```

```python
tele_num_spaces = '123 456 7890'
```

```python
m = re.match(pattern='\d{10}', string=tele_num_spaces)
print(m)
#패턴을 못찾음
```

```python
if m:
    print('match')
else:
    print('no match')
```

```python
p = '\d{3}\s?\d{3}\s?\d{4}'
m = re.match(pattern=p, string=tele_num_spaces)
print(m)
# \s?를 넣어 패턴 다시 만듦
```

```python
tele_num_space_paren_dash = '(123) 456-7890'
p = '\(?\d{3}\)?\s?\d{3}\s?-?\d{4}'
m = re.match(pattern=p, string=tele_num_space_paren_dash)
print(m)
# (지역코드)전화-번호
```

```python
cnty_tele_num_space_paren_dash = '+1 (123) 456-7890'
p = '\+?1\s?\(?\d{3}\)?\s?\d{3}\s?-?\d{4}'
m = re.match(pattern=p, string=cnty_tele_num_space_paren_dash)
print(m)
#국가코드까지 있는 경우
```

<compile 메서드로 정규식 메서드 사용하기>

```python
p = re.compile('\d{10}') #패턴 컴파일 -> 반복사용 가능
s = '1234567890'
m = p.match(s)
print(m)
```
