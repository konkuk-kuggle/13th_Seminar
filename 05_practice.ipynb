# 데이터프레임에 시리즈 연결하기

```python
new_row_series = pd.Series(['n1', 'n2', 'n3', 'n4'])
```

```python
print(pd.concat([df1, new_row_series]))
```

# 행 1개로 구성된 데이터프레임 생성하여 연결하기

```python
new_row_df = pd.DataFrame([['n1', 'n2', 'n3', 'n4']], columns=['A', 'B', 'C', 'D'])
print(new_row_df)
```

```python
print(pd.concat([df1, new_row_df]))
```

```python
# 예전 버전에서만 가능
print(df1.append(new_row_df))
```

```python
data_dict = {'A': 'n1', 'B': 'n2', 'C': 'n3', 'D': 'n4'}
print(df1.append(data_dict, ignore_index=True))
```

# 다양한 방법으로 데이터 연결하기

```python
# ignore_index 인자 사용하기
row_concat_i = pd.concat([df1, df2, df3], ignore_index=True)
print(row_concat_i)
```

```python
# 열 방향으로 데이터 연결하기
col_concat = pd.concat([df1, df2, df3], axis=1)
print(col_concat)
```

```python
print(col_concat['A'])
```

```python
col_concat['new_col_list'] = ['n1', 'n2', 'n3', 'n4']
print(col_concat)
```

```python
print(pd.concat([df1, df2, df3], axis=1, ignore_index=True))
```

# 공통 열과 공통 인덱스만 연결하기

```python
df1.columns = ['A', 'B', 'C', 'D']
df2.columns = ['E', 'F', 'G', 'H']
df3.columns = ['A', 'C', 'F', 'H']
print(df1)
print(type(df1))

print(df2)
print(type(df2))

print(df3)
print(type(df3))
```

```python
row_concat = pd.concat([df1, df2, df3])
print(row_concat)
```

```python
print(pd.concat([df1, df2, df3], join='inner'))
```

```python
print(pd.concat([df1,df3], ignore_index=False, join='inner'))
```

```python
df1.index = [0, 1, 2, 3]
df2.index = [4, 5, 6, 7]
df3.index = [0, 2, 5, 7]

print(df1)

print(df2)

print(df3)
```

```python
col_concat = pd.concat([df1, df2, df3], axis=1)
print(col_concat)
```

```python
print(pd.concat([df1, df3], axis=1, join='inner'))
```

# merge 메서드 사용하기

```python
person = pd.read_csv('/content/drive/MyDrive/KUGGLE/데이터 분석을 위한 판다스 입문/data/survey_person.csv')
site = pd.read_csv('/content/drive/MyDrive/KUGGLE/데이터 분석을 위한 판다스 입문/data/survey_site.csv')
survey = pd.read_csv('/content/drive/MyDrive/KUGGLE/데이터 분석을 위한 판다스 입문/data/survey_survey.csv')
visited = pd.read_csv('/content/drive/MyDrive/KUGGLE/데이터 분석을 위한 판다스 입문/data/survey_visited.csv')

print(person, '\n')
print(site, '\n')
print(visited, '\n')
print(survey)
```

```python
visited_subset = visited.loc[[0, 2, 6], ]
```

```python
o2o_merge = site.merge(visited_subset, left_on='name', right_on='site')
print(o2o_merge)
```

```python
m2o_merge = site.merge(visited, left_on='name', right_on='site')
print(m2o_merge)
```

```python
ps = person.merge(survey, left_on='ident', right_on='person')
vs = visited.merge(survey, left_on='ident', right_on='taken')

print(ps)

print(vs)
```

```python
ps_vs = ps.merge(vs, left_on=['ident', 'taken', 'quant', 'reading'], right_on=['person', 'ident', 'quant', 'reading'])
```

```python
print(ps_vs.loc[0, ])
```
