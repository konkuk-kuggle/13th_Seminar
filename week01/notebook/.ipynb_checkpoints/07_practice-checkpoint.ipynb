# 1개의 열만 고정하고 나머지 열을 행으로 바꾸기

```python
import pandas as pd
pew = pd.read_csv('../data/pew.csv')
print(pew.head())
print(pew.iloc[:, 0:6])
```

```python
#melt 메서드 : 지정한 열의 데이터를 모두 행으로 정리
pew_long = pd.melt(pew, id_vars='religion') #id_vars : 위치를 그대로 유지할 열의 이름
print(pew_long.head())

#var_name : value_vars로 위치를 변경한 열의 이름, value_name : var_name으로 위치를 변경할 열의 데이터를 저장한 열의 이름
pew_long = pd.melt(pew, id_vars='religion', var_name='income', value_name='count')
print(pew_long.head())
```

# 2개 이상의 열을 고정하고 나머지 열을 행으로 바꾸기

```python
billboard = pd.read_csv('../data/billboard.csv')

print(billboard.iloc[0:5, 0:16])
```

```python
#5개의 행 고정
billboard_long = pd.melt(billboard, id_vars=['year', 'artist', 'track', 'time', 'date.entered'], var_name='week', value_name='rating')
print(billboard_long.head())
```

# ebola 데이터 집합 살펴보기

```python
ebola = pd.read_csv('../data/country_timeseries.csv')
print(ebola.columns)
print(ebola.iloc[:5, [0, 1, 2, 3, 10, 11]])

ebola_long = pd.melt(ebola, id_vars=['Date', 'Day']) #Date, Day 고정하여 피벗
print(ebola_long.head())
```

# 열 이름 나누고 데이터 프레임에 추가하기

```python
variable_split = ebola_long.variable.str.split('_') #'_' 문자를 기준으로 분리
print(variable_split[:5]) #Cases_Guinea -> Cases, Guinea
print(type(variable_split))
print(type(variable_split))

status_values = variable_split.str.get(0) #발병, 죽음 (Cases)
country_values = variable_split.str.get(1) #나라 이름 (Guinea)
print(status_values[:5])
print(status_values[-5:])
print(country_values[:5])
print(country_values[-5:])

#분리한 문자열로 데이터프레임에 추가
ebola_long['status'] = status_values 
ebola_long['country'] = country_values
print(ebola_long.head())
```

# concat 메서드를 응용하여 데이터프레임에 열 추가하기

```python
variable_split = ebola_long.variable.str.split('_', expand=True) 
variable_split.columns = ['status', 'country'] 
ebola_parsed = pd.concat([ebola_long, variable_split], axis=1) #concat() : 분리한 데이터를 바로 데이터프레임에 추가

print(ebola_parsed.head())
```

# 기상 데이터의 여러 열을 하나로 정리하기

```python
weather = pd.read_csv('../data/weather.csv') 
print(weather.iloc[:5, :11])
```

```python
weather_melt = pd.melt(weather, id_vars=['id', 'year', 'month', 'element'], var_name='day', value_name='temp') 
print(weather_melt.head()) #일별 온도(d1~...) 피벗

weather_tidy = weather_melt.pivot_table( #pivot_table : 행과 열의 위치를 바꿔 정리
    index=['id', 'year', 'month', 'day'], #위치를 유지할 열 이름
    columns='element', #피벗할 열 이름
    values='temp' #새로운 열의 데이터가 될 열 이름
)
print(weather_tidy)
```

```python
weather_tidy_flat = weather_tidy.reset_index() #데이터프레임 인덱스를 reset_index로 새로 지정(0~)
print(weather_tidy_flat.head())
```
