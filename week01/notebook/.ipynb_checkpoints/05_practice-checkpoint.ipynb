# concat 메서드 사용하기

```python
import pandas as pd

df1=pd.read_csv('../data/concat_1.csv')
df2=pd.read_csv('../data/concat_2.csv')
df3=pd.read_csv('../data/concat_3.csv')
```

```python
row_concat=pd.concat([df1,df2,df3]) #연결한 데이터프레임 반환
print(row_concat)
print(row_concat.iloc[3,]) #행 데이터 추출
```

# 데이터프레임에 시리즈 연결하기

```python
new_row_series=pd.Series(['n1','n2','n3','n4']) #데이터프레임에 시리즈 추가
print(pd.concat([df1, new_row_series]))
```

# 행 1개로 구성된 데이터프레임 생성하여 연결하기

```python
new_row_df=pd.DataFrame([['n1','n2','n3','n4']], columns=['A','B','C','D'])
#print(df1)
print(new_row_df) #1개의 행을 가지는 데이터 프레임 생성
print(pd.concat([df1, new_row_df])) #df1에 연결

#연결할 데이터프레임이 1개라면 append 메서드 사용 가능 (현재 append 사용 불가)
#print(df1.append(new_row_df))
```

```python
#append 메서드 + 딕셔너리
#data_dict={'A': 'n1', 'B': 'n2', 'C': 'n3', 'D': 'n4'}
#print(df1.append(data_dict, ignore_index=True))
```

# 다양한 방법으로 데이터 연결하기

```python
print(f"df1:\n{df1}\ndf2:\n{df2}\ndf3:\n{df3}\n")

row_concat_i=pd.concat([df1, df2, df3], ignore_index=True) #ignore_index : 데이터 연결 후 데이터프레임의 인덱스를 0부터 다시 지정
print(row_concat_i)
```

```python
#열 방향 데이터 연결
col_concat=pd.concat([df1, df2, df3], axis=1) #axis=0 : 행 방향, axis=1 : 열 방향
print(col_concat)
print(col_concat['A']) #해당 열 이름(A)의 데이터 모두 추출

col_concat['new_col_list']=['n1','n2','n3','n4'] #새로운 열 추가
print(col_concat) 

print(pd.concat([df1, df2, df3], axis=1, ignore_index=True)) #열 이름 다시 지정(0부터)
```
