# 앤스콤 데이터 집합 불러오기

```python
import seaborn as sns #seaborn : 4개의 서로 다른 데이터셋이 하나의 데이터프레임에 합쳐져 있는 구조

anscombe = sns.load_dataset("anscombe")
print(f'{anscombe}\n') #데이터셋이 로마 숫자로 분류
print(type(anscombe))
```

# 한 번에 4개의 그래프 그리기

```python
#데이터셋 분류
dataset_1=anscombe[anscombe['dataset']=='I']
dataset_2=anscombe[anscombe['dataset']=='II']
dataset_3=anscombe[anscombe['dataset']=='III']
dataset_4=anscombe[anscombe['dataset']=='IV']

fig=plt.figure() #캔버스 생성

#기본틀에 격자 추가 (행 크기, 열 크기, 격자 번호)
axes1=fig.add_subplot(2,2,1)
axes2=fig.add_subplot(2,2,2)
axes3=fig.add_subplot(2,2,3)
axes4=fig.add_subplot(2,2,4)

#그래프 그리기
axes1.plot(dataset_1['x'], dataset_1['y'], 'o') #'o' : marker (원형 점으로 표기)
axes2.plot(dataset_2['x'], dataset_2['y'], 'o')
axes3.plot(dataset_3['x'], dataset_3['y'], 'o')
axes4.plot(dataset_4['x'], dataset_4['y'], 'o')

#그래프 제목 추가
axes1.set_title("dataset_1")
axes2.set_title("dataset_2")
axes3.set_title("dataset_3")
axes4.set_title("dataset_4")

fig.suptitle("Anscombe Data")

fig.tight_layout() #글자 안겹치게
```

# 기초 그래프 그리기 - 히스토그램, 산점도, 박스 그래프

```python
#데이터 불러오기
tips=sns.load_dataset("tips")
print(tips.head())
print(type(tips))
```

```python
#히스토그램
fig=plt.figure()
axes1=fig.add_subplot(1,1,1)

axes1.hist(tips['total_bill'], bins=10) #bins=x축 간격 지정
axes1.set_title('Histogram of Total Bill')
axes1.set_xlabel('Frequency')
axes1.set_ylabel('Total Bill')
```

```python
#산점도
scatter_plot=plt.figure()
axes1=scatter_plot.add_subplot(1,1,1)

axes1.scatter(tips['total_bill'], tips['tip']) #x축, y축
axes1.set_title('Scatterplot of Total Bill vs Tip')
axes1.set_xlabel('Total Bill')
axes1.set_ylabel('Tip')
```

```python
#박스 그래프
boxplot=plt.figure()
axes1=boxplot.add_subplot(1,1,1)

axes1.boxplot([tips[tips['sex']=='Female']['tip'],
               tips[tips['sex']=='Male']['tip']],
               #labels=['Female', 'Male'])
               tick_labels=['Female', 'Male']) #3.9버전부터 파라미터 이름 변경
              

axes1.set_xlabel('Sex')
axes1.set_ylabel('Tip')
axes1.set_title('Boxplot of Tips by Sex')
```

# 단변량 그래프 그리기 - 히스토그램

```python
import seaborn as sns

tips=sns.load_dataset("tips")

ax=plt.subplots()
ax=sns.distplot(tips['total_bill'])
ax.set_title('Total Bill Histogram with Density Plot')
```

```python
#밀집도 그래프 제외
ax=plt.subplots()
ax=sns.distplot(tips['total_bill'], kde=False) #밀집도 함수를 제외하고 싶다면 kde=False
ax.set_title('Total Bill Histogram')
ax.set_xlabel('Total Bill')
ax.set_ylabel('Frequency')
```

```python
#히스토그램 제외
ax=plt.subplots()
ax=sns.distplot(tips['total_bill'], hist=False) #밀집도 함수만 나타내고 싶다면 hist=False
ax.set_title('Total Bill Density')
ax.set_xlabel('Total Bill')
ax.set_ylabel('Unit Probability')
```

```python
#양탄자 그래프 추가
#양탄자 그래프 : 데이터값이 어디에 몰려있는 지 나타냄
ax=plt.subplots()
ax=sns.distplot(tips['total_bill'], rug=True) #양탄자 그래프 나타내고 싶다면 rug=True
ax.set_title('Total Bill Histogram with Density and Rug Plot')
ax.set_xlabel('Total Bill')
```

```python
#count 그래프
ax=plt.subplots()
ax=sns.countplot(x='day', data=tips) #축 데이터 명확히 제시
ax.set_title('Count of dats')
ax.set_xlabel('Day of the Week')
ax.set_ylabel('Frequency')
```

# 다변량 그래프 그리기

```python
#바이올린 그래프
ax=plt.subplots()
ax=sns.violinplot(x='time', y='total_bill', hue='sex', data=tips, split=True)
```

```python
#산점도, 관계 그래프
scatter=sns.lmplot(x='total_bill', y='tip', data=tips, hue='sex', fit_reg=False)
fig=sns.pairplot(tips, hue='sex')
```

```python
#산점도 그래프 조절
#점 크기 조절
scatter=sns.lmplot(x='total_bill', y='tip', data=tips, fit_reg=False, hue='sex',
                   scatter_kws={'s': 50}) #점 크기 조절시 scatter_kws에 딕셔너리 형태로 인잣값 전달 #'s' : 점의 크기(size)

#점을 다른 기호로 표현
scatter=sns.lmplot(x='total_bill', y='tip', data=tips, fit_reg=False, hue='sex',
                   markers=['o','x'], scatter_kws={'s': 50})

#4개 그래프 한 번에 그리기
anscombe_plot=sns.lmplot(x='x', y='y', data=anscombe, fit_reg=False) #fit_reg : 회귀선

#4개 그래프 나누어 그리기
anscombe_plot=sns.lmplot(x='x', y='y', data=anscombe, fit_reg=False,
                         col='dataset', col_wrap=2) #col : 데이터 그룹을 구분할 열, col_wrap : 그래프를 그릴 열의 최대값
```

# 데이터프레임과 시리즈로 그래프 그리기

```python
#plot속성의 hist 메서드로 히스토그램 그리기
ax=plt.subplots()
ax=tips['total_bill'].plot.hist()
```

```python
#투명도 조절
fig, ax=plt.subplots() #2개의 시리즈 선택
ax=tips[['total_bill', 'tip']].plot.hist(alpha=0.5, bins=20, ax=ax) #alpha : 투명도
```

```python
#밀집도 그래프
ax=plt.subplots()
ax=tips['tip'].plot.kde()

#산점도 그래프
fig, ax=plt.subplots()
ax=tips.plot.scatter(x='total_bill', y='tip', ax=ax)

#육각 그래프
#fig, ax= plt.subplots()
fig, (ax1, ax2)=plt.subplots(1, 2) #육각그래프 나란히 출력
ax1=tips.plot.hexbin(x='total_bill', y='tip', ax=ax1)
ax2=tips.plot.hexbin(x='total_bill', y='tip', gridsize=10, ax=ax2) #육각형 크기 변경

#박스 그래프
fig, ax=plt.subplots()
ax=tips.plot.box(ax=ax)
```

```python
ax=tips.plot.hexbin(x='total_bill', y='tip', gridsize=10, ax=ax) #육각형 크기 변경

#박스 그래프
fig, ax=plt.subplots()
ax=tips.plot.box(ax=ax)
```

# 알아두면 좋아요

```python
#FacetGrid 클래스로 그룹별 그래프 그리기
facet=sns.FacetGrid(tips, col='time') #'time'을 기준으로 그룹을 나눔
facet.map(sns.distplot, 'total_bill', rug=True)

#다변량 그래프
facet=sns.FacetGrid(tips, col='day', hue='sex')
facet=facet.map(plt.scatter, 'total_bill', 'tip')
facet=facet.add_legend() #색으로 구분(hue)하는 데이터가 있을 떄, 각 데이터가 무엇을 의미하는 지

#더 많은 변수
facet=sns.FacetGrid(tips, col='time', row='smoker', hue='sex') #time=Dinner,Lunch / smok
```
