# 문자열 추출하기

```python
word = 'grail'
sent = 'a scratch'

print(word[0])
print(sent[0])
print(word[0:3])
print(sent[-1])
print(sent[-9:-8])
print(sent[0:-8])
```

# 전체 문자열 추출하기

```python
print(sent[2:-1]) #전체 문자열 추출 시 음수 사용 x
print(sent[-7:-1])
s_len = len(sent)
print(s_len) #len을 이용하여 추출
print(sent[2:s_len])
```

# join, splitlines, replace 메서드 실습하기

```python
#join : 문자열 연결
d1 = '40°' 
m1 = "46'" 
s1 = '52.837"' 
u1 = 'N'

d2 = '73°' 
m2 = "58'" 
s2 = '26.302"' 
u2 = 'W'

coords = ' '.join([d1, m1, s1, u1, d2, m2, s2, u2]) #join() 앞 문자 지정 시, 해당 문자를 단어 사이에 넣어 연결
print(coords)
```

```python
#splitlines : 여러 행을 가진 문자열을 분리 후 리스트로 반환
multi_str = """Guard: What? Ridden on a horse?
King Arthur: Yes!
Guard: You're using coconuts!
King Arthur: What?
Guard: You've got ... coconut[s] and you're bangin' 'em together. 
""" 
print(multi_str)

multi_str_split = multi_str.splitlines() #줄바꿈 기호를 기준으로 분리(행 별)
print(multi_str_split)

guard = multi_str_split[::2]
print(guard)
```

```python
#replace : 문자열 치환
guard = multi_str.replace("Guard: ", "").splitlines()[::2] 
print(guard)
```

# 문자열 포매팅 실습하기

```python
var = 'flesh wound' 
s = "It's just a {}!" #단어를 삽입할 위치에 {} 지정

print(s.format(var)) #format()로 원하는 단어 전달
print(s.format('scratch'))

#여러 단어 전달 시 인덱스 이용
s = """Black Knight: 'Tis but a {0}.
King Arthur: A {0}? Your arm's off!
""" 
print(s.format('scratch'))

s = 'Hayden Planetarium Coordinates: {lat}, {lon}' 
print(s.format(lat='40.7815° N', lon='73.9733° W'))
```

# 수치값 포매팅 실습

```python
print('Some digits of pi: {}'.format(3.14159265359))
print("In 2005, Lu Chao of China recited {:,} digits of pi".format(67890)) #3자리마다 , 찍기
print("I remember {0:.4} or {0:.4%} of what Lu Chao recited".format(7/67890)) #유효 숫자 4자리(0이 아닌 수가 나오는 지점 부터), 소수점 밑 4자리
print("My ID number is {0:05d}".format(42)) #5자리 숫자, 빈칸은 0으로 채우
```

# % 연산자로 포매팅하기

```python
s = 'I only know %d digits of pi' % 7 #10진수 : %d
print(s)
print('Some digits of %(cont)s: %(value).2f' % {'cont': 'e', 'value': 2.718}) #문자열 : %s, #실수 : %f
```

# 정규식으로 전화번호 패턴 찾기

```python
import re #정규표현

tele_num = '1234567890'

m = re.match(pattern='\d\d\d\d\d\d\d\d\d\d', string=tele_num) #숫자 10자리(패턴)가 맞는지 확인
print(type(m)) #매칭된 결과 객
print(m) #어디서부터 어디까지 어떤 글자가 매칭되었는지 요약
print(bool(m)) #매칭 성공 시 True, 매칭 실패 시 False

if m:
    print('match') 
else:
    print('no match')
    
print(m.start()) #매칭이 시작된 인덱스 번호
print(m.end()) #매창이 끝난 바로 다음 인덱스 번호
print(m.span()) #시작과 끝 인덱
print(m.group()) #매칭된 문자열
```

```python
tele_num_spaces = '123 456 7890'
m = re.match(pattern='\d{10}', string=tele_num_spaces) #패턴 일치(공백)
print(m)

if m:
    print('match') 
else:
    print('no match')
```

```python
p = '\d{3}\s?\d{3}\s?\d{4}' #\s : 공백, ? : 0번 또는 1번 발
m = re.match(pattern=p, string=tele_num_spaces) 
print(m)

tele_num_space_paren_dash = '(123) 456-7890' 
p = '\(?\d{3}\)?\s?\d{3}\s?-?\d{4}' 
m = re.match(pattern=p, string=tele_num_space_paren_dash) 
print(m)

cnty_tele_num_space_paren_dash = '+1 (123) 456-7890' 
p = '\+?1\s?\(?\d{3}\)?\s?\d{3}\s?-?\d{4}' #+ : 최소 한 번 이상 반복
m = re.match(pattern=p, string=cnty_tele_num_space_paren_dash) 
print(m)
```

# compile 메서드로 정규식 메서드 사용하기

```python
p = re.compile('\d{10}') #규칙을 객체로 만들어둠
s = '1234567890' 
m = p.match(s) 
print(m)
```
