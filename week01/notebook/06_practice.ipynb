# 누락값 확인하기

```python
from numpy import nan
```

```python
print(nan==True)
print(nan==False)
print(nan==0)
print(nan=='')

print(nan==nan) #누락값은 값 자체가 없기 때문에 자기 자신과 비교해도 False 출력
```

```python
#누락값 확인
import pandas as pd
print(pd.isnull(nan))

print(pd.notnull(nan))
print(pd.notnull(42))
print(pd.notnull('missing'))
```

# 누락값을 포함한 데이터를 불러올 때

```python
visited=pd.read_csv('../data/survey_visited.csv')
survey=pd.read_csv('../data/survey_survey.csv')
print(visited)
print(survey)
```

# 데이터 집합을 연결할 때 누락값이 발생하는 경우

```python
vs=visited.merge(survey, left_on='ident', right_on='taken') #데이터집합 연결
print(vs)
```

# 데이터를 입력할 때 누락값이 발생하는 경우

```python
num_legs=pd.Series({'goat':4, 'amoeba': nan})
print(num_legs)
print(type(num_legs))

scientists=pd.DataFrame({
    'Name': ['Rosaline Franklin', 'William Gosset'],
    'Occupation': ['Chemist', 'Statistician'],
    'Born': ['1920-07-25', '1876-06-13'],
    'Died': ['1958-04-16', '1937-10-16'],
    'missing': [nan, nan]})
print(scientists)
print(type(scientists))
```

# 인덱스를 다시 만들 때 누락값이 발생하는 경우

```python
gapminder=pd.read_csv('../data/gapminder.tsv', sep='\t')

life_exp=gapminder.groupby(['year'])['lifeExp'].mean() #lifeExp 열의 평균
print(life_exp)

#print(life_exp.loc[range(2000, 2010), ]) #존재하지 않는 인덱스 => 오
print(life_exp.reindex(range(2000, 2010), )) #reindex() : 누락값 nan으로 채워짐
y2000=life_exp[life_exp.index>2000] #불린 추출(조건식)
print(y2000)
```

# 누락값의 개수 구하기

```python
ebola=pd.read_csv('../data/country_timeseries.csv')

print(ebola.count()) #누락값이 아닌 개수 구하기

num_rows=ebola.shape[0] #shape[0]에 전체 행의 데이터 개수가 저장
num_missing=num_rows-ebola.count() #전체 데이터 개수 - 누락값이 아닌 값의 개수
print(num_missing)
```

```python
import numpy as np
print(np.count_nonzero(ebola.isnull()))
#ebola.isnull() : 결측값이면 True, 결측값이 아니면 False
#np.count_nonzero() : True값 카운팅

print(np.count_nonzero(ebola['Cases_Guinea'].isnull()))
```

```python
#value_counts : 지정한 열의 빈도
#dropna=False : 결측값을 포함하여 측정(NaN)
print(ebola.Cases_Guinea.value_counts(dropna=False).head()) #Cases_Guinea 열의 누락값 개수
```

# 누락값을 다른 값으로 변경하기

```python
print(ebola.fillna(0).iloc[0:10, 0:5]) #fillna(0) : 누락값을 0으로 변경
#print(ebola.fillna(method='ffill').iloc[0:10, 0:5]) #method='ffill' : 누락값을 이전(누락값이 아닌 직전 행)값으로 변경
#print(ebola.fillna(method='bfill').iloc[0:10, 0:5]) #method='bfill' : 누락값을 이후(누락값이 아닌 직후 행)값으로 변경

#현재 방식
print(ebola.ffill().iloc[0:10, 0:5])
print(ebola.bfill().iloc[0:10, 0:5])

print(ebola.interpolate().iloc[0:10, 0:5]) #누락값 양쪽의 중간값으로 변경
```

# 누락값 삭제하기

```python
print(ebola.shape)

#ebola.dropna() : 결측값이 있는 행 삭제
ebola_dropna=ebola.dropna()
print(ebola_dropna.shape)

print(ebola_dropna)
```

# 누락값이 포함된 데이터 계산하기

```python
ebola['Cases_multiple']=ebola['Cases_Guinea']+ebola['Cases_Liberia']+ebola['Cases_SierraLeone']

ebola_subset=ebola.loc[:, ['Cases_Guinea', 'Cases_Liberia', 'Cases_SierraLeone', 'Cases_multiple']]
print(ebola_subset.head(n=10)) #누락값이 하나라도 있는 행은 계산결과가 0 (Cases_multiple)

print(ebola.Cases_Guinea.sum(skipna=True)) #누락값 무시하여 계산
print(ebola.Cases_Guinea.sum(skipna=False))
```
