# 데이터프레임 만들기

```python
#1
scientists=pd.DataFrame({
    'Name':['Rosaline Franklin', 'William Gosset'],
    'Occupation':['Chemist', 'Statistician'],
    'Born':['1920-07-25', '1876-06-13'],
    'Died':['1958-04-16', '1937-10-16'],
    'Age':[37, 61]}
)
print(scientists)
#2 - 인덱스 지정
scientists=pd.DataFrame(
    data={
        'Occupation':['Chemist', 'Statistician'],
        'Born':['1920-07-25', '1876-06-13'],
        'Died':['1958-04-16', '1937-10-16'],
        'Age':[37, 61]},
    index=['Rosaline Franklin', 'William Gosset'],
    columns=['Occupation', 'Born', 'Age', 'Died']
)
print(scientists)

#3 - 딕셔너리 순서 보장
from collections import OrderedDict

scientists=pd.DataFrame(OrderedDict([
    ('Name', ['Rosaline Franklin', 'William Gosset']),
    ('Occupation', ['Chemist', 'Statistician']),
    ('Born', ['1920-07-25', '1876-06-13']),
    ('Died', ['1958-04-16', '1937-10-16']),
    ('Age', [37, 61])
])
)
print(scientists)
```

# 데이터프레임에서 시리즈 선택하기

# 시리즈의 mean, min, max, std 메서드 사용하기

```python
ages=scientists['Age']
print(ages)

print(ages.mean()) #평균
print(ages.min()) #최솟값
print(ages.max()) #최대값
print(ages.std()) #표준편차
```

# 시리즈와 불린 추출 사용하기

```python
scientists=pd.read_csv('../data/scientists.csv')

ages=scientists['Age']
print(ages.max())
print(ages.mean())

print(ages[ages > ages.mean()]) #True인 값만
print(ages > ages.mean()) #각 원소마다 비교 -> T/F로 이루어진 Series
manual_bool_values=[True, True, False, False, True, True, False, True] #True인 인덱스 값만
print(ages[manual_bool_values])
```

# 시리즈와 브로드캐스팅

```python
#브로드캐스팅 : 서로 크기(차원)가 다른 배열끼리 연산을 가능하게 해주는 규칙
#크기가 같은 벡터
print(ages+ages)
print(ages*ages)

#벡터 +* 스칼라
print(ages+100)
print(ages*2)

#크기가 다른 벡터
print(pd.Series([1, 100])) #인덱스 2개
print(ages+pd.Series([1,100])) #ages : 인텍스8개 => 인덱스가 일치하는 인덱스 0, 1만 계산

#벡터와 벡터의 연산은 일치하는 인덱스끼리 수행
print(ages)
rev_ages=ages.sort_index(ascending=False) #인덱스 역순
#sort_index: 인덱스 기준으로 정렬 #ascending=False : 내림차순
print(rev_ages)

print(ages*2)
print(ages+rev_ages) # =ages*2
```

# 열의 자료형 바꾸기와 새로운 열 추가하기

```python
#자료형 바꾸기
print(scientists['Born'].dtype)
print(scientists['Died'].dtype)

#자료형->datetime
born_datetime=pd.to_datetime(scientists['Born'], format='%Y-%m-%d')
print(born_datetime)
died_datetime=pd.to_datetime(scientists['Died'], format='%Y-%m-%d')
print(died_datetime)

#새로운 열 추가
scientists['born_dt'], scientists['died_dt']=(born_datetime, died_datetime) #'born_dt', 'died_dt' 열 추가
print(scientists.head())
print(scientists.shape) #행:8, 열:7(5->7로 증가)
```
